```python
import os
import math
from PIL import Image
import matplotlib.pyplot as plt
```

```python
from dotenv import load_dotenv
load_dotenv()

api_key = os.getenv("RF_API_KEY")

# if u run this code in colab comment the above lines and uncomment the below lines and add your api key
# !pip install -U ultralytics roboflow 
# api_key = "your_api_key_here"
```

```python
from roboflow import Roboflow
rf = Roboflow(api_key=api_key)
project = rf.workspace("haya-e5lla").project("mini_project-1cddt")
version = project.version(3)
dataset = version.download("yolov8-obb")
```

```python
from ultralytics import YOLO
model = YOLO("yolov8n-obb.pt")
```

```python
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=50,
    imgsz=800,
    batch=8,
    lr0=0.001,
    patience=30
)
```

```python
metrics = model.val()
```

```python
pred = model.predict(
    source=f"{dataset.location}/test/images",
    save=True,
    conf=0.25
)
```

```python
result_dir = "runs/obb/predict"
images = [f for f in os.listdir(result_dir) if f.endswith((".jpg", ".jpeg", ".png"))]

cols = 4
rows = math.ceil(len(images) / cols)

fig, axes = plt.subplots(rows, cols, figsize=(18, 4*rows))

if rows == 1:
    axes = [axes] if cols == 1 else axes
axes = axes.ravel()

for i, img_name in enumerate(images):
    img = Image.open(os.path.join(result_dir, img_name))
    axes[i].imshow(img)
    axes[i].axis("off")

for j in range(i + 1, len(axes)):
    axes[j].axis("off")

fig.tight_layout()

display(fig)
plt.close(fig)
```
